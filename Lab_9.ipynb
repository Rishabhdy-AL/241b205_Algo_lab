```python
def dijkstra(graph, start):

    distances = {}
    predecessors = {}
    visited = []

    for node in graph:
        distances[node] = float('inf')
        predecessors[node] = None

    distances[start] = 0

    while len(visited) < len(graph):

        min_node = None
        min_distance = float('inf')

        # Find the unvisited node with the smallest distance
        for node in graph:
            if node not in visited and distances[node] < min_distance:
                min_distance = distances[node]
                min_node = node

        # If no unvisited node is reachable, break
        if min_node is None:
            break

        visited.append(min_node)

        # Update distances of neighbors
        for neighbor, weight in graph[min_node].items():
            new_distance = distances[min_node] + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                predecessors[neighbor] = min_node

    return distances, predecessors
```

# Shortest Path Using Dijkstra Algorithm in Python
Problem Description:
Given a weighted graph with non negative edge weights. Find the shortest distance from a source vertex to all other vertices using Dijkstra algorithm

#Theory
Dijkstra algorithm finds shortest distance from one source node to all other nodes in a weighted graph with non negative weights. Dictionary structure stores nodes as keys and neighbor nodes with weights as values.

##Algorithm Steps

1. Represent graph using dictionary.
2. Set distance of every node as infinity.
3. Set distance of source node as 0.
4. Select node with smallest distance among unvisited nodes.
5. Update distance of neighbor nodes.
6. Repeat until all nodes get visited.

```python
def dijkstra(graph, start):

    distances = {}
    visited = []

    for node in graph:
        distances[node] = float('inf')

    distances[start] = 0

    while len(visited) < len(graph):

        min_node = None
        min_distance = float('inf')

        for node in graph:
            if node not in visited and distances[node] < min_distance:
                min_distance = distances[node]
                min_node = node

        visited.append(min_node)

        for neighbor, weight in graph[min_node].items():
            new_distance = distances[min_node] + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance

    return distances
```

```python
graph = {
'A': {'B': 4, 'C': 2},
'B': {'A': 4, 'C': 5, 'D': 10},
'C': {'A': 2, 'B': 5, 'D': 3},
'D': {'B': 10, 'C': 3}
}

result = dijkstra(graph, 'A')

for node in result:
    print("Distance from A to", node, "=", result[node])
```

```python
def get_user_graph():
    graph = {}
    num_nodes = int(input("Enter the number of nodes in the graph: "))

    for i in range(num_nodes):
        node = input(f"Enter name for node {i+1}: ").strip().upper()
        graph[node] = {}

    print("\nNow, enter the connections and weights. Type 'done' when finished with a node.")
    for node in graph:
        while True:
            connection_input = input(f"Enter a connection for node '{node}' (e.g., 'neighbor_node:weight') or 'done': ").strip()
            if connection_input.lower() == 'done':
                break
            try:
                neighbor, weight_str = connection_input.split(':')
                neighbor = neighbor.strip().upper()
                weight = int(weight_str.strip())

                if neighbor not in graph:
                    print(f"Warning: Node '{neighbor}' not defined. Adding it to the graph with no outgoing edges for now.")
                    graph[neighbor] = {}

                if weight < 0:
                    print("Error: Edge weight cannot be negative. Please enter a non-negative integer.")
                    continue
                graph[node][neighbor] = weight
            except ValueError:
                print("Invalid input format. Please use 'neighbor:weight' (e.g., 'B:5').")
            except Exception as e:
                print(f"An unexpected error occurred: {e}")
    return graph

def reconstruct_path(predecessors, start_node, end_node):
    path = []
    current = end_node
    while current is not None:
        path.insert(0, current)
        current = predecessors[current]
    if path[0] == start_node:
        return path
    return []
```

```python
print("\n--- Demonstrating Shortest Paths and Distances ---")

# Using the previously defined graph for demonstration
graph = {
    'A': {'B': 4, 'C': 2},
    'B': {'A': 4, 'C': 5, 'D': 10},
    'C': {'A': 2, 'B': 5, 'D': 3},
    'D': {'B': 10, 'C': 3}
}
start_node = 'A'

distances, predecessors = dijkstra(graph, start_node)

print(f"\nShortest distances from source node '{start_node}':")
for node, dist in distances.items():
    print(f"  To {node}: {dist}")

print(f"\nShortest paths from source node '{start_node}':")
for node in graph:
    path = reconstruct_path(predecessors, start_node, node)
    if path:
        print(f"  Path to {node}: {' -> '.join(path)} (Distance: {distances[node]}) ")
    else:
        print(f"  Path to {node}: Not reachable")
```
